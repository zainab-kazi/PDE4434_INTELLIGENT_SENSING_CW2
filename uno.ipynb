```python
import cv2
import os

# Set your input and output folders
input_folder = r"C:\Users\Zainab\UNO_final\dataset\blue"  # Change for other colors
output_folder = r"C:\Users\Zainab\UNO_final\templates"

# Create output folder if it doesn't exist
os.makedirs(output_folder, exist_ok=True)

# Collect all image files
image_files = [f for f in os.listdir(input_folder) if f.endswith(('.jpg', '.png'))]

print(f"Found {len(image_files)} images. Starting cropping...")

for filename in image_files:
    img_path = os.path.join(input_folder, filename)
    img = cv2.imread(img_path)

    if img is None:
        print(f"Could not read {filename}, skipping.")
        continue

    # Resize for easier manual cropping
    img = cv2.resize(img, (0, 0), fx=0.1, fy=0.1)

    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    print(f"Cropping: {filename} — Select ROI and press ENTER or SPACE")

    # Show image and wait briefly to ensure window is visible
    cv2.imshow("Crop", gray)
    cv2.waitKey(1)  # Wait a moment to display the window

    # Select ROI
    roi = cv2.selectROI("Crop", gray, showCrosshair=True)

    if roi == (0, 0, 0, 0):
        print(f"❌ No ROI selected for {filename}, skipping.")
        cv2.destroyWindow("Crop")
        continue

    # Crop the selected region
    template = gray[int(roi[1]):int(roi[1]+roi[3]), int(roi[0]):int(roi[0]+roi[2])]

    # Extract base name and save with prefix
    base_label = os.path.splitext(filename)[0]
    save_name = f"t_{base_label}.jpg"
    save_path = os.path.join(output_folder, save_name)

    cv2.imwrite(save_path, template)
    print(f"✅ Saved template: {save_path}")

    # Close the crop window
    cv2.destroyWindow("Crop")

# Destroy any remaining windows after loop ends
cv2.destroyAllWindows()
```

```python
import os

# Root folders
image_root = r"C:/Users/Zainab/UNO_final/uno_yolo/data"
label_root = r"C:/Users/Zainab/UNO_final/uno_yolo/labels"  # Save labels here now

# Define class mappings
class_map = {
    "red": 0,
    "blue": 1,
    "yellow": 2,
    "green": 3,
    "wild": 4
}

# Aliases to recognize as wild cards
alias_map = {
    "color_change": "wild",
    "plus_4": "wild"
}

# Loop through subfolders like 'train', 'val', 'test'
for subfolder in os.listdir(image_root):
    image_dir = os.path.join(image_root, subfolder)
    if not os.path.isdir(image_dir):
        continue

    label_dir = os.path.join(label_root, subfolder)
    os.makedirs(label_dir, exist_ok=True)

    for filename in os.listdir(image_dir):
        if filename.endswith(".jpg") or filename.endswith(".png"):
            filename_lower = filename.lower()

            # Try to find the color
            color = None
            for c in class_map:
                if c in filename_lower:
                    color = c
                    break

            # Try alias if not found
            if not color:
                for alias, mapped_color in alias_map.items():
                    if alias in filename_lower:
                        color = mapped_color
                        break

            if not color:
                print(f"Skipping {filename}, no known color found in name")
                continue

            class_id = class_map[color]
            yolo_line = f"{class_id} 0.5 0.5 0.9 0.9\n"

            # Save .txt file in the correct data/<subfolder>/ directory
            txt_filename = os.path.splitext(filename)[0] + ".txt"
            txt_path = os.path.join(label_dir, txt_filename)

            with open(txt_path, "w") as f:
                f.write(yolo_line)

            print(f"[{subfolder}] Saved label: {txt_filename}")
```

```python
## CODE WITHOUT GUI AND DIRECT WEBCAM FOR UNO DETECTION
import cv2
import numpy as np
import os

# === 1. HSV Color Ranges ===
hsv_ranges = {
    "red": [
        (np.array([0, 100, 100]), np.array([10, 255, 255])),
        (np.array([160, 100, 100]), np.array([180, 255, 255])),
    ],
    "blue": [(np.array([100, 150, 50]), np.array([140, 255, 255]))],
    "green": [(np.array([40, 100, 50]), np.array([80, 255, 255]))],
    "yellow": [(np.array([20, 100, 100]), np.array([30, 255, 255]))],
}

# === 2. Template Loading ===
def load_templates(template_folder):
    template_data = []
    for fname in os.listdir(template_folder):
        if fname.startswith("t_") and fname.endswith(".jpg"):
            color_label = fname[2:].split('.')[0]  # strip "t_" and extension
            template_path = os.path.join(template_folder, fname)
            temp_img = cv2.imread(template_path, cv2.IMREAD_GRAYSCALE)
            if temp_img is None:
                continue

            color_detected = None
            for known_color in hsv_ranges:
                if known_color in color_label.lower():
                    color_detected = known_color
                    break

            template_data.append({
                "label": color_label,
                "template": temp_img,
                "w": temp_img.shape[1],
                "h": temp_img.shape[0],
                "color": color_detected if color_detected else "wild"
            })
    print(f"✅ Loaded {len(template_data)} templates.")
    return template_data

# === 3. Detect Card Color Using HSV ===
def detect_card_color(card_img):
    hsv = cv2.cvtColor(card_img, cv2.COLOR_BGR2HSV)
    color_scores = {}

    for color, ranges in hsv_ranges.items():
        mask = np.zeros(hsv.shape[:2], dtype=np.uint8)
        for lower, upper in ranges:
            mask |= cv2.inRange(hsv, lower, upper)
        color_scores[color] = cv2.countNonZero(mask)

    detected_color = max(color_scores, key=color_scores.get)
    return detected_color if color_scores[detected_color] > 500 else None

# === 4. Match Template From ROI ===
def match_card_template(roi_gray, templates):
    best_match = None
    best_val = 0
    for temp in templates:
        template = temp["template"]
        label = temp["label"]
        if template.shape[0] > roi_gray.shape[0] or template.shape[1] > roi_gray.shape[1]:
            continue
        res = cv2.matchTemplate(roi_gray, template, cv2.TM_CCOEFF_NORMED)
        _, max_val, _, _ = cv2.minMaxLoc(res)
        if max_val > best_val:
            best_val = max_val
            best_match = label
    return best_match if best_val > 0.45 else None

# === 5. Webcam Detection Loop ===
def start_webcam_detection(template_data):
    cap = cv2.VideoCapture(0)  # Use 0 for the default camera
    if not cap.isOpened():
        print("❌ Could not open webcam.")
        exit()

    last_detected_label = None  # Variable to store the last detected card label

    while True:
        ret, frame = cap.read()
        if not ret:
            break

        # Flip the frame horizontally (mirror effect)
        frame = cv2.flip(frame, 1)

        output = frame.copy()
        blurred = cv2.GaussianBlur(frame, (5, 5), 0)
        gray = cv2.cvtColor(blurred, cv2.COLOR_BGR2GRAY)
        edged = cv2.Canny(gray, 50, 150)

        contours, _ = cv2.findContours(edged, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

        detected_label = None  # Reset detected label for the current frame

        for cnt in contours:
            area = cv2.contourArea(cnt)
            if area < 1000 or area > 100000:
                continue

            x, y, w, h = cv2.boundingRect(cnt)
            card_roi = frame[y:y+h, x:x+w]
            card_gray = gray[y:y+h, x:x+w]

            # === Template Match ===
            template_match = match_card_template(card_gray, template_data)
            if template_match:
                parts = template_match.split('_')
                color = parts[0] if parts[0] in hsv_ranges else None
                symbol = '_'.join(parts[1:]) if len(parts) > 1 else "unknown"

                # If template has no color, try detecting it
                if color is None:
                    color = detect_card_color(card_roi)

                if color and symbol:
                    detected_label = f"{color}_{symbol}_card"
                    last_detected_label = detected_label  # Update last detected label
                    cv2.rectangle(output, (x, y), (x + w, y + h), (0, 255, 0), 2)
                    cv2.putText(output, detected_label, (x, y - 10), cv2.FONT_HERSHEY_SIMPLEX,
                                0.7, (0, 255, 0), 2)

        # Display the last detected label if no card is detected in the current frame
        if last_detected_label and not detected_label:
            cv2.putText(output, f"Last Detected: {last_detected_label}", (10, 30),
                        cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 2)

        # === Display Output ===
        cv2.imshow("UNO Card Detection", output)

        key = cv2.waitKey(1) & 0xFF
        if key == 27 or key == ord('q'):
            break

    cap.release()
    cv2.destroyAllWindows()

# === Main Execution ===
if __name__ == "__main__":
    template_folder = r"C:\Users\Zainab\UNO_final\templates"
    template_data = load_templates(template_folder)
    start_webcam_detection(template_data)
```

```python
## CODE WITH GUI AND WEBCAM OPTION FOR UNO DETECTION
import cv2
import numpy as np
import os
from tkinter import Tk, Button, Label

# === 1. HSV Color Ranges ===
hsv_ranges = {
    "red": [
        (np.array([0, 100, 100]), np.array([10, 255, 255])),
        (np.array([160, 100, 100]), np.array([180, 255, 255])),
    ],
    "blue": [(np.array([100, 150, 50]), np.array([140, 255, 255]))],
    "green": [(np.array([40, 100, 50]), np.array([80, 255, 255]))],
    "yellow": [(np.array([20, 100, 100]), np.array([30, 255, 255]))],
}

# === 2. Template Loading ===
def load_templates(template_folder):
    template_data = []
    for fname in os.listdir(template_folder):
        if fname.startswith("t_") and fname.endswith(".jpg"):
            color_label = fname[2:].split('.')[0]  # strip "t_" and extension
            template_path = os.path.join(template_folder, fname)
            temp_img = cv2.imread(template_path, cv2.IMREAD_GRAYSCALE)
            if temp_img is None:
                continue

            color_detected = None
            for known_color in hsv_ranges:
                if known_color in color_label.lower():
                    color_detected = known_color
                    break

            template_data.append({
                "label": color_label,
                "template": temp_img,
                "w": temp_img.shape[1],
                "h": temp_img.shape[0],
                "color": color_detected if color_detected else "wild"
            })
    print(f"✅ Loaded {len(template_data)} templates.")
    return template_data

# === 3. Detect Card Color Using HSV ===
def detect_card_color(card_img):
    hsv = cv2.cvtColor(card_img, cv2.COLOR_BGR2HSV)
    color_scores = {}

    for color, ranges in hsv_ranges.items():
        mask = np.zeros(hsv.shape[:2], dtype=np.uint8)
        for lower, upper in ranges:
            mask |= cv2.inRange(hsv, lower, upper)
        color_scores[color] = cv2.countNonZero(mask)

    detected_color = max(color_scores, key=color_scores.get)
    return detected_color if color_scores[detected_color] > 500 else None

# === 4. Match Template From ROI ===
def match_card_template(roi_gray, templates):
    best_match = None
    best_val = 0

    for temp in templates:
        template = temp["template"]
        label = temp["label"]
        if template.shape[0] > roi_gray.shape[0] or template.shape[1] > roi_gray.shape[1]:
            continue
        res = cv2.matchTemplate(roi_gray, template, cv2.TM_CCOEFF_NORMED)
        _, max_val, _, _ = cv2.minMaxLoc(res)
        if max_val > best_val:
            best_val = max_val
            best_match = label

    return best_match if best_val > 0.5 else None  # Adjusted threshold

# === 5. Resize While Keeping Aspect Ratio ===
def resize_with_aspect_ratio(image, max_width, max_height):
    h, w = image.shape[:2]
    aspect_ratio = w / h

    if w > max_width or h > max_height:
        if aspect_ratio > 1:  # Wider than tall
            new_w = max_width
            new_h = int(max_width / aspect_ratio)
        else:  # Taller than wide
            new_h = max_height
            new_w = int(max_height * aspect_ratio)
        resized = cv2.resize(image, (new_w, new_h))

        # Center the image by adding padding
        canvas = np.zeros((max_height, max_width, 3), dtype=np.uint8)
        y_offset = (max_height - new_h) // 2
        x_offset = (max_width - new_w) // 2
        canvas[y_offset:y_offset+new_h, x_offset:x_offset+new_w] = resized
        return canvas
    return image

# === 6. Process Webcam ===
def process_webcam_frame(template_data):
    cap = cv2.VideoCapture(0)
    if not cap.isOpened():
        print("❌ Unable to access the webcam.")
        return

    while True:
        ret, frame = cap.read()
        if not ret:
            print("❌ Failed to capture frame from webcam.")
            break

        # Flip the frame horizontally for a mirrored effect
        frame = cv2.flip(frame, 1)

        # Resize the frame while keeping aspect ratio
        frame = resize_with_aspect_ratio(frame, 800, 600)
        output = frame.copy()
        blurred = cv2.GaussianBlur(frame, (5, 5), 0)
        gray = cv2.cvtColor(blurred, cv2.COLOR_BGR2GRAY)
        edged = cv2.Canny(gray, 50, 150)

        contours, _ = cv2.findContours(edged, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

        detected_any_card = False  # Flag to check if any card has been detected

        for cnt in contours:
            area = cv2.contourArea(cnt)
            if area < 1000 or area > 200000:  # Adjusted thresholds for better detection
                continue

            x, y, w, h = cv2.boundingRect(cnt)
            card_roi = frame[y:y+h, x:x+w]
            card_gray = gray[y:y+h, x:x+w]

            # Template Matching
            template_match = match_card_template(card_gray, template_data)
            if template_match:
                detected_any_card = True
                parts = template_match.split('_')
                color = parts[0] if parts[0] in hsv_ranges else None
                symbol = '_'.join(parts[1:]) if len(parts) > 1 else "unknown"

                # If template has no color, try detecting it
                if color is None:
                    color = detect_card_color(card_roi)

                # Correctly label the detected card
                label = f"{color}_{symbol}_card" if color and symbol else "unknown_card"
                print(f"✅ Detected Card: {label}")
                cv2.rectangle(output, (x, y), (x + w, y + h), (0, 255, 0), 2)
                cv2.putText(output, label, (x, y - 10), cv2.FONT_HERSHEY_SIMPLEX,
                            0.7, (0, 255, 0), 2)

        if not detected_any_card:
            print("❌ No UNO card detected in the frame.")

        # Display the frame with bounding boxes
        cv2.imshow("Webcam UNO Detection", output)

        # Exit on pressing 'q'
        if cv2.waitKey(1) & 0xFF == ord('q'):
            break

    cap.release()
    cv2.destroyAllWindows()

# === 7. GUI ===
def launch_gui(template_data):
    root = Tk()
    root.title("UNO Card Detection")
    Label(root, text="UNO Card Detection", font=("Helvetica", 16)).pack(pady=10)

    Button(root, text="Use Webcam", command=lambda: process_webcam_frame(template_data), width=20, height=2).pack(pady=10)

    Label(root, text="Press 'q' to exit webcam", font=("Helvetica", 12)).pack(pady=10)
    root.mainloop()

# === Main Execution ===
if __name__ == "__main__":
    template_folder = r"C:\Users\Zainab\UNO_final\templates"
    template_data = load_templates(template_folder)
    launch_gui(template_data)
```
